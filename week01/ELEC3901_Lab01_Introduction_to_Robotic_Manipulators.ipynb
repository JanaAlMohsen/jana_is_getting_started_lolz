### Does the order matter?



*The same translation and rotation, multiplied in the two orders. Left: `transl2 @ trot2` moves to (0.15, -0.10),
    then turns 30 degrees in place - that is `se2`. Right: `trot2 @ transl2` turns the axes 30 degrees about {B} first, then moves 0.15 and
    -0.10 **along the turned axes**, so {A} ends up at (0.180, -0.012).*

```python
T1 = transl2(0.15, -0.10) @ trot2(np.radians(30))     # your Task 3
T2 = trot2(np.radians(30)) @ transl2(0.15, -0.10)
print('transl2 @ trot2 puts {A} at', np.round(T1[:2, 2], 4), '   trot2 @ transl2 puts it at', np.round(T2[:2, 2], 4))
SE2().plot(frame='B', color='k', length=0.05)
SE2(T1).plot(frame='1', color='b', length=0.05); SE2(T2).plot(frame='2', color='r', length=0.05)
plt.axis('equal'); plt.grid(); plt.show()
```
